```python
import pandas as pd
import common_mod_1 as cmm
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import cross_validate
```

```python
adult_census = cmm.dfFromArffFile('/data/phpMawTba.arff')
adult_census.drop(columns=['education-num', 'fnlwgt'], inplace=True)
adult_census.columns
```

```python
features, target = adult_census[adult_census.columns.difference(['class'])], adult_census["class"]
features_num = features.select_dtypes(exclude='object')
features_num.columns
```

```python
model_pipe = make_pipeline(StandardScaler(), LogisticRegression())
model_pipe
```

```python
cv = cross_validate(model_pipe, features_num, target, cv=5)
"""
fit_time: the time to train the model on the training data for each fold.
score_time: the time to predict with the model on the testing data for each fold.
test_score: the default score on the testing data for each fold.
"""
cv
```

Mean accuracy and variation of the accuracy across folds.

```python
print(f'mean: {cv["test_score"].mean():.2f}, standard_dev: {cv["test_score"].std():.3f}')
```
