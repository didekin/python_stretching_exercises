```python
from sklearn.neighbors import KNeighborsClassifier as KNN
from inria_course_ml import common_mod as cmm
```

```python
adult_census = cmm.dfFromArffFile('/data/phpMawTba.arff')
adult_census.head(10)
```

```python
census_types = adult_census.dtypes
census_types
```

```python
target_name = "class"
target = adult_census[target_name].astype("category").str.decode('utf-8')
target
```

```python
features_num = census_types[census_types != 'object']
features_num_names = sorted(features_num.index)
features_num_names.remove('fnlwgt')
features_num_Df = adult_census[features_num_names]
features_num_Df.columns
```

```python
features_num_Df
```

```python
model = KNN()
_ = model.fit(features_num_Df, target)
```

```python
target_predicted = model.predict(features_num_Df)
(target_predicted == target).mean()
```

The score method is the same as comparing actual class with predicted class. I do not see a big difference with KNN between using test or training data in the score method.

```python
model.score(features_num_Df, target)
```
