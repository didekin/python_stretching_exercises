```python
import os

import pandas as pd
import seaborn as sn
```

```python
file_path = os.path.abspath('../') + '/data/Auto.csv'
autos = pd.read_csv(file_path)
print(autos.head())
```

```python
types_series = autos.dtypes
types_series
```

Categorical and numeric variables

```python
categorial_vars = types_series[types_series == "object"]
print(categorial_vars.index[0])
categorial_vars.count()
```

```python
numeric_vars = types_series[types_series != "object"]
numeric_vars.index
```

```python
count_names = autos["name"].value_counts()
count_names
```

```python
_ = autos[numeric_vars.index].hist(figsize=(22, 16))
```

```python
cols_plot = numeric_vars.index
sample_size = 200
target = 'mpg'
_ = sn.pairplot(data=autos[:sample_size],vars=cols_plot, hue=target, height=3, plot_kws={"alpha": 0.5}, diag_kind="hist", diag_kws={"bins": 10})
```
