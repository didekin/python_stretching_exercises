```python
import numpy as np
import matplotlib.pyplot as plot
import os as o
```

```python
fig1, ax1 = plot.subplots(figsize=(8, 8))
rng = np.random.default_rng(11)
x1 = rng.standard_normal(100)
y1 = rng.standard_normal(100)
ax1.scatter(x1, y1)
ax1.set_title('Scatter plot');
```

```python
fig2, ax2 = plot.subplots(figsize=(8, 9))
x2 = rng.standard_normal(50)
y2 = rng.standard_normal(50)
ax2.set_title('Standard plot')
ax2.plot(x2, y2);
```

```python
fig2.set_size_inches(12,6)
fig2
```

```python
fig3,ax3 = plot.subplots(nrows=2, ncols=2, figsize=(15,6))
ax3[0,1].plot(x2,y2, 'o')
ax3[0,0].plot(x1,y1);
```
