```python
import numpy as np
import pandas as pd
autos = pd.read_csv('../data/Auto.csv')
print(type(autos))
print(f'dimensions: {autos.shape}')
autos
```

SELECTING ROWS AND COLUMNS OF A DATAFRAME

Indexes for columns:

```python
autos.columns
```

Indexes for rows: automatic and ad-hoc.

```python
autos.index
```

```python
autos_index_col = autos.set_index('name')
autos_index_col
```

The variable 'name' is not already present in the indexes for columns:

```python
autos_index_col.columns
```

Selecting rows: the three first rows.

```python
autos[:3]
```

Selecting by the index of rows (names):

```python
print(autos_index_col.index)
rows_ex = ['ford torino', 'chevrolet impala']
autos_index_col.loc[rows_ex]
```

Selecting rows by a list of numbers:

```python
autos_index_col.iloc[[3,4]]
```

```python
autos.iloc[[3,4]]
```

A boolean filter

```python
filter_horsepow = autos['horsepower'] > 215
autos[filter_horsepow]
```

Selecting columns:

```python
horsepow_var = autos['horsepower']
print(type(horsepow_var))
print(horsepow_var)
print('sum' , sum(horsepow_var))
```

```python
np.unique(horsepow_var)
```

```python
autos[['mpg', 'horsepower']]
```

Selecting columns by a list of numbers:

```python
autos_index_col.iloc[:, [1,2]]
```

Selecting a combination of rows and columns:

```python
autos.iloc[[2,3],[1,2]]
```

A combination of index row and column names:

```python
autos_index_col.loc[['chevrolet impala', 'ford torino'], ['cylinders', 'displacement']]
```

Selecting rows with lambda and a list of columns.

```python
autos_index_col.loc[lambda df: df['cylinders'] == 8, ['weight','displacement']]
```

```python
autos_index_col.loc[lambda df: (df['cylinders'] == 8) & (df.index.str.contains('ford') |
                               df.index.str.contains('datsun')), ['weight', 'displacement', 'cylinders']]
```
