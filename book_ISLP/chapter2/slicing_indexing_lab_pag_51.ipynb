SUBMATRICES: THREE WAYS

```python
M = m_4_4
M[[1,3]][:,[0,2]]
```

```python
idx = np.ix_([1,3],[0,2])
M[idx]
```

```python
M[1:4:2,0:3:2]
```

FILTERING ROWS AND COLUMNS WITH BOOLEANS

```python
M
```

```python
filter_rows = np.zeros(M.shape[0], bool)
filter_rows[[1,2]] = True
print(type(filter_rows))
filter_rows
```

Filtering rows

```python
M[np.array([False, True, True, False])]
```

```python
M[filter_rows]
```

Difference with filtering with an np_array: numbers in the list are dealt as indexes of the rows to be selected.

```python
M[np.array([0,1,1,0])]
```

Filtering columns

```python
M[:, [0,3]]
```

```python
filter_cols = np.zeros(M.shape[1], bool)
filter_cols[[0,3]] = True
M[:, filter_cols]
```

Filtering rows and columns

```python
filter_row_col = np.ix_(filter_rows, filter_cols)
M[filter_row_col]
```
