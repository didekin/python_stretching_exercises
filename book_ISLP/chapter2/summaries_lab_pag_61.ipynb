```python

import pandas as pd
from matplotlib.pyplot import subplots

autos = pd.read_csv('../data/Auto.csv')
```

Plot by accessing directly the columns in a data frame.

```python
fig1, ax1 = subplots(figsize=(8, 8))
ax1.plot(autos['horsepower'], autos['mpg'], 'o');
```

Plot by using the data frame property 'plot':

```python
ax2 = autos.plot.scatter('horsepower', 'mpg')
ax2.set_title('Horsepower vs. MPG')
ax2.figure.savefig('plots/horsepower_mileage.png')
```

Mixing the data frame 'plot' property and the 'subplots()' method.

```python
fig3, ax3 = subplots(ncols=3, figsize=(15, 5))
autos.plot.scatter('horsepower', 'mpg', ax=ax3[1]);
```

Accessing columns by the attribute 'name':

```python
autos.horsepower
```

Transforming a quantitative variable into a qualitative one:

```python
autos.cylinders.dtype
```

```python
autos.cylinders = pd.Series(autos.cylinders, dtype='category')
print(f'cylinders type = {autos.cylinders.dtype}')
autos.cylinders
```

```python
fig4, ax4 = subplots()
autos.boxplot('mpg', by='cylinders', ax=ax4);
```

An histogram.

```python
fig5, ax5 = subplots()
autos.hist('mpg', color='red', bins=10, ax=ax5);
```

The  pd.plotting.scatter_matrix() function:

```python
pd.plotting.scatter_matrix(autos, figsize=(9, 11));
```

```python
pd.plotting.scatter_matrix(autos[['acceleration', 'mpg', 'weight']], figsize=(9, 7));
```

The describe() method produces a numerical summary of each column in a data frame.

```python
autos[['acceleration', 'weight', 'mpg']].describe()
```

```python
autos[['cylinders']].describe()
```
