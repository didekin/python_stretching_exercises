```python
import numpy as np
import pandas as pd
```

Generating a matrix with a 25% of nan values.

```python
rng = np.random.default_rng(3)
dim = (30, 3)
A = rng.standard_normal(dim)
M = rng.choice([0, np.nan], p=[0.75, .25], size=A.shape)
AM = A + M
AM = pd.DataFrame(AM, columns=['col_1','col_2','col_3'])
AM
```

Printing the % NaN values by column:

```python
for col in AM.columns:
    print(f'{col} has {np.isnan(AM[col]).mean():.2%} of NaN values')
```
